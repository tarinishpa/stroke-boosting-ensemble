8.Write a Python function that accepts an integer as input
and determines whether the number is prime.


```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
def prime_sum(a,b):
  sum=0
  for i in range(a,b+1):
    flag=1
    if i<2:
      continue
    for j in range (2,i):
      if i%j==0:
        flag=0
    if flag==1:
      sum+=i
  print(sum)
prime_sum(1,20)

```

19.Write a Python program to find the sum of the digits of a
number repeatedly until a single digit is obtained.

```python
n = int(input("Enter a number: "))

while n >= 10:
    s = 0
    while n > 0:
        digit = n % 10
        s = s + digit
        n = n // 10
    n = s

print("Sum =", n)
```

32.Write a Python program to merge multiple dictionaries.
If a key appears in more than one dictionary, add the
corresponding values.

```python
d1={'a':10}
d2={'a':5, 'b':20}

result = {}

for d in (d1, d2):
    for key in d:
        if key in result:
            result[key] += d[key]
        else:
            result[key] = d[key]

print(result)
```

33.Write a Python program to analyze student course
enrollment records and determine the course with the
highest number of enrolled students.

```python
lst=[{'Sabbir':['AI','DB']},{'Tanvir':['AI','ML']},{'Nafisa':['AI']}]
course_count={}
for d in lst:
  for student in d:
    for course in d[student]:
      if course in course_count:
        course_count[course]+=1
      else:
        course_count[course]=1

mx = max(course_count, key=lambda x: course_count[x])

print("Course:", mx)
print("Total Enrollment:", course_count[mx])
```

42.Write a Python program to validate a password based on
the following conditions: it must contain at least one
uppercase letter, at least one digit, one special character,
and no more than five lowercase letters.

```python
pas=input("Enter password:")
up=0
digit=0
sp=0
lw=0
for c in pas:
  if c.isalpha() and c.isupper():
    up+=1
  elif c.isdigit():
    digit+=1
  elif c.isalpha() and c.islower():
    lw+=1
  else:
    sp+=1
if up>=1 and digit>=1 and lw<=5 and sp==1:
  print("Valide password")
else:
  print("Not valide password")
```
